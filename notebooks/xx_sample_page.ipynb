## Covariance Formation in FMCW MIMO Radar

In an FMCW MIMO radar, direction-of-arrival estimation is normally performed after range and Doppler processing. The processing flow for one radar frame can be represented as

$$
\text{ADC data}
\rightarrow
\text{Range FFT}
\rightarrow
\text{Doppler FFT}
\rightarrow
\text{Range--Doppler data cube}
\rightarrow
\text{Covariance estimation}
\rightarrow
\text{DoA estimation}.
$$

After range and Doppler processing, each range–Doppler cell contains one complex sample for every physical or virtual antenna channel. For a virtual array with $M$ channels, the spatial observation associated with range bin $r$ and Doppler bin $d$ is

$$
\mathbf{x}[r,d]
=

\begin{bmatrix}
X_0[r,d] \
X_1[r,d] \
\vdots \
X_{M-1}[r,d]
\end{bmatrix}
\in
\mathbb{C}^{M}.
$$

The vector $\mathbf{x}[r,d]$ is one **spatial snapshot** of the antenna array. Its elements contain the amplitudes and phases measured across the virtual antenna channels for the selected range–Doppler cell.

A covariance matrix computed from only this observation is

$$
\hat{\mathbf{R}}_{xx}
=

\mathbf{x}[r,d]\mathbf{x}^{H}[r,d].
$$

Because it is formed from a single snapshot, this matrix has rank at most one:

$$
\operatorname{rank}
\left(
\mathbf{x}[r,d]\mathbf{x}^{H}[r,d]
\right)
\leq 1.
$$

A rank-one covariance matrix cannot generally represent several independent spatial sources, even when multiple targets contribute to the same range–Doppler cell. It is also poorly conditioned and cannot be directly inverted for Capon beamforming without regularization.

Therefore, covariance-based DoA estimation requires several spatial observations:

$$
\mathbf{X}
=

\begin{bmatrix}
\mathbf{x}_0 &
\mathbf{x}*1 &
\cdots &
\mathbf{x}*{L-1}
\end{bmatrix},
$$

from which the sample covariance matrix is estimated as

$$
\hat{\mathbf{R}}_{xx}
=

\frac{1}{L}
\mathbf{X}\mathbf{X}^{H}.
$$

Within one FMCW MIMO radar frame, additional snapshots can be formed using several approaches:

1. spatial vectors from neighboring Doppler cells at the same range;
2. spatial vectors from a local range–Doppler neighborhood;
3. overlapping subarrays of the physical or virtual antenna array.

Each approach increases the number of observations available for covariance estimation, but introduces a different trade-off between covariance stability, spatial aperture, and the risk of combining signals from different targets.
